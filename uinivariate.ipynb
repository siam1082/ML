##EDA

Primary Goals of EDA

Understand Data Structure: Inspecting row/column counts, data types, and previewing entries (e.g., df.shape, df.info(), df.head()).

Spot Missing Data & Outliers: Identifying null values (df.isnull().sum()) and extreme values that could distort models (using boxplots, histograms, or IQR metrics).

Feature Relationships & Correlations: Measuring how variables interact (e.g., df.corr(), scatter plots, heatmaps) to select relevant predictors.

Data Distribution: Checking whether numerical variables follow a normal distribution, are skewed, or have multiple modes.

Formulate Hypotheses: Guiding feature engineering, data cleaning choices (imputation, encoding, scaling), and model selection.

3. Boxplot

Key Components Explained25th Percentile ($Q_1$ / First Quartile):The value below which $25\%$ of the sorted observations fall. It forms the lower boundary (left edge) of the central box.Median ($Q_2$ / 50th Percentile):The exact midpoint of the sorted dataset. Exactly $50\%$ of values lie below it and $50\%$ lie above it. Unlike the mean, the median is resistant to extreme values.75th Percentile ($Q_3$ / Third Quartile):The value below which $75\%$ of the sorted data falls. It marks the upper boundary (right edge) of the central box.Interquartile Range (IQR):The distance covering the central $50\%$ of the data distribution:$$\text{IQR} = Q_3 - Q_1$$Whiskers / Fences (Acceptable Range):Data thresholds that determine the boundaries for typical observations:$$\text{Lower Fence} = Q_1 - 1.5 \times \text{IQR}$$$$\text{Upper Fence} = Q_3 + 1.5 \times \text{IQR}$$Outliers:Any observation situated strictly outside the calculated fences:$$\text{Value} < \text{Lower Fence} \quad \text{or} \quad \text{Value} > \text{Upper Fence}$$In visual plots, these appear as isolated points or markers beyond the whisker lines.
