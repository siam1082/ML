```python
import numpy as np
```

O D tensors


```python

arr = np.array(5)

print(arr)
print(arr.ndim)
print(arr.shape)
```

1-D vector


```python
a = np.array([1,2,3])

print(a)
print(a.ndim)
print(a.shape)
```

2-D Matrix


```python


a = np.array([
    [1, 2],
    [3, 4]
])

print(a)
print(a.ndim)
print(a.shape)
```

3-D Tensor

```python
a = np.array([
    [
        [1,2],
        [3,4]
    ],
    [
        [4,5],
        [23,45]
    ]
])

print(a)
print(a.ndim)
print(a.shape)
```
