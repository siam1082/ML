Confidence Interval (CI) হলো অনুমানের এমন একটা রেঞ্জ (পরিসীমা), যার মধ্যে আসল সত্য মানটি (True Population Value) থাকার প্রবল সম্ভাবনা থাকে।

সহজ কথায়: তুমি যখন পুরো দুনিয়ার সব ডেটা না নিয়ে কেবল একটা ছোট স্যাম্পল (নমুনা) নিয়ে কাজ করো, তখন তোমার বের করা গড় (Mean) কখনোই ১০০% নির্ভুল হবে না। তাই একটা নির্দিষ্ট সংখ্যা না বলে, ভুলের ঝুঁকি মাথায় রেখে যে সম্ভাব্য সীমানা উল্লেখ করা হয়, সেটাই Confidence Interval।

বাস্তব উদাহরণ (Example)
ধরা যাক, তুমি একটি ক্যাম্পাসের সকল শিক্ষার্থীর গড় ওজন জানতে চাও।

ক্যাম্পাসে মোট শিক্ষার্থী ১০,০০০ জন। সবার ওজন মাপা অসম্ভব।

তুমি র‍্যান্ডমলি ৫০ জন শিক্ষার্থীর ওজন মেপে গড় পেলে ৬৫ কেজি।

এখন তুমি জোর দিয়ে বলতে পারবে না যে পুরো ১০,০০০ জনের গড় ঠিক ৬৫.০ কেজিই হবে।

তখন স্ট্যাটিস্টিক্স হিসাব করে তুমি বললে:

"আমি ৯৫% আত্মবিশ্বাসী যে পুরো ক্যাম্পাসের আসল গড় ওজন ৬৩ কেজি থেকে ৬৭ কেজির মধ্যে আছে।"

এখানে:

৬৫ কেজি = বারের উচ্চতা (Sample Mean)।

৬৩ থেকে ৬৭ কেজি = Confidence Interval (মাথার ওপরের কালো দাগটি)।

# 3. Boxplot

```python
sns.boxplot(data=titanic, x="Sex", y="Age",hue = "Survived")
```

4. distplot (numerical - catagorcial)

```python
sns.displot(data=titanic, x="Age", hue="Sex", kde=True, multiple = 'dodge')
```

```python
# kde plot
sns.kdeplot(data=titanic, x="Age", hue="Sex", fill=True, alpha=0.3)
plt.show()
```

# 5. Heatmap (catagorical - catagorical )

```python
sns.heatmap(pd.crosstab(titanic["Pclass"], titanic["Survived"]))
```

# 6. Cluster map

```python
# ১. Pclass এবং Survived-এর ক্রস-ট্যাব তৈরি
ct = pd.crosstab(titanic["Pclass"], titanic["Survived"])

# ২. ক্লাস্টার ম্যাপ তৈরি
sns.clustermap(ct, annot=True, fmt="d", cmap="Blues")
plt.show()
```
