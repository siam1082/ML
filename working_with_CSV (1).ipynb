The primary purpose of squeeze is dimensionality reduction—removing redundant 1-dimensional axes to convert a data structure into its simplest underlying format.

```python
import pandas as pd

# ১টি কলাম usecols দিয়ে ফিল্টার করে .squeeze('columns') কল করুন
series = pd.read_csv(
    'data.csv',
    encoding='utf-8-sig',
    usecols=['Year']
).squeeze('columns')

print(type(series))  # <class 'pandas.core.series.Series'>
series.head()
```

```python
# Skip the first 2 rows of the file
df = pd.read_csv('data.csv', skiprows=2)

# Skip specific row indices (e.g., row 0, 2, and 5)
df = pd.read_csv('data.csv', skiprows=[0, 2, 5])

# Skip using a lambda function (e.g., skip all even-indexed rows)
df = pd.read_csv('data.csv', skiprows=lambda x: x % 2 == 0)

df
```

```python
# Read only the first 100 rows
df = pd.read_csv('data.csv', nrows=100)
df
```

```python
# Skip corrupted lines instead of crashing
df = pd.read_csv('data.csv', on_bad_lines='skip')
df
```

The converters parameter lets you apply custom Python functions to specific columns while the CSV is being read.

```python


# Define transformation functions
def clean_currency(val):
    return float(val.replace('$', '').replace(',', ''))

def clean_name(val):
    return val.strip().title()

# Apply converters during load
df = pd.read_csv(
    'data.csv',
    converters={
        'Salary': clean_currency,
        'Full_Name': clean_name
    }
)
df
```
